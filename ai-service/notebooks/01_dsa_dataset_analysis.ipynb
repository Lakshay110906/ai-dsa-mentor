```python
pd.crosstab(df["topic"],df["difficulty"])
```

```python
df['desc_words'] = df['description'].str.split().str.len()
df[['title','desc_words']].head()
```

```python
df['desc_words'].describe()
```

```python
import matplotlib.pyplot as plt
fig,axes = plt.subplots(1,2,figsize=(12,4))

df['topic'].value_counts().plot(kind='bar',ax=axes[0],color='steelblue')
axes[0].set_title("Problems per topic")
axes[0].set_ylabel("Count")
df['difficulty'].value_counts().plot(kind='bar',ax=axes[1],color='orange')
axes[1].set_title("Problems per difficulty")
axes[1].set_ylabel("Count")

plt.tight_layout()
plt.show()
```

There are 35 problems total and every topic has same no of problems ie balanced 
In Problems per difficulty graph Medium problems are more than easy and hard one
Avg description length is about 19 words
